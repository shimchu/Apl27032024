```python
def matrix_multiply(matrix1, matrix2):
  '''
  Defining the function matrix_multiply taking in two matrices and multiplying them.
  Raises  ValueErrors if the dimensions of the matrices make them incompatible,
  Raises TypeError if the elements in the matrices are not numbers (int, float, complex)
  Returns the Final output matrix, if the two matrices given as input are valid
  '''
  if not matrix1 or not matrix2:                                    # Checking if any matrix is empty (to raise ValueError), in Python - empty lists are evaluated as false.
      raise ValueError("One or both matrices are empty")

  if len(matrix1[0]) != len(matrix2):                               # Checking if the no. of columns in the first Matrix (A) is not equal to no. of rows in second Matrix (B)
    if len(matrix2[0]) == len(matrix1):                             # If A's columns and B's rows aren't equal in number, but the vice versa is true then we swap the matrices, esssentially outputting B x A
      matrix1, matrix2 = matrix2, matrix1
    else:
      raise ValueError("Incompatible matrices for multiplication")  # Else we raise a ValueError as multiplication can't be computed either way (A x B or B x A)



  def are_all_elements_numeric(matrix):
    ''' Function to check if all elements in a matrix are numbers'''
    return all(isinstance(element, (int, float, complex)) for row in matrix for element in row)

  if (not are_all_elements_numeric(matrix1)) or (not are_all_elements_numeric(matrix2)):    # Checking if the Function outputs False, then it raises Type Error
      raise TypeError("A matrix has non numeric values as elements")



  m = len(matrix1)      # Number of rows in First Matrix
  n = len(matrix1[0])   # Number of columns in First Matrix = Number of rows in Second Matrix
  p = len(matrix2[0])   # Number of columns in Second Matrix

  if any(len(row) != n for row in matrix1) or any(len(row) != p for row in matrix2):        # Checking if all rows have the same length in each matrix (This helps check if empty lists as elements other than the first)
      raise ValueError("Rows in a matrix have inconsistent lengths")


  C = [[0 for _ in range(p)] for _ in range(m)]   # Initializing the Output Matrix with size m x p and all elements as 0.

  for i in range(m):                              # Iterating through the rows of matrix 1
    for j in range(p):                            # Iterating through the columns of matrix 2, for a particular Row of Matrix 1
      for k in range(n):                          # Iterating through the elements of both matrices, for a particular row of Matrix 1 and column of Matrix 2
        C[i][j] += matrix1[i][k] * matrix2[k][j]  # Adding up the product of the corresponding elements and equating it as the corresponding element in C



  return C                                        # returning C as the output of the function
```
